```python
import pandas as pd
```

```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/content/drive/MyDrive/Toyota.csv', index_col=0, na_values=['??', '????'])
df
```

```python
# Removing missing values from the dataframe
df.dropna(axis=0, inplace=True)
df
```

```python
# scatter plot
plt.scatter(df['Age'], df['Price'], c = 'red')
plt.show()
```

```python
# Histogram is a graphical representation of data using bars of different height.
plt.hist(df['KM'])
```

```python
plt.hist(df['KM'], color = 'green', edgecolor='white', bins=5)
```

```python
counts = [979,120,12]
FuelType=('Petrol', 'Diesel', 'CNG')
index=np.arange(len(FuelType)) # Corrected: Use len(FuelType) instead of len('FuelType')
plt.bar(index, counts, color=['red', 'blue', 'cyan'])
plt.xticks(index, FuelType) # Add x-axis labels
plt.title('Fuel Type Distribution') # Add a title for clarity
plt.xlabel('Fuel Type') # Add x-axis label
plt.ylabel('Number of Cars') # Add y-axis label
plt.show()
```
