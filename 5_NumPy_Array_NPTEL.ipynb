```python
list1= [1,2,3,4,5,6]
print(list1)
```

```python
# To make an numpy array from a list.
import numpy as np
```

```python
arr = np.array(list1, dtype = int)
print(arr)
```

```python
print(type(arr))
print(len(arr))
# no. of axis of an array.
print(arr.ndim)
print(arr.shape) # 6 rows, 1 column
```

```python
# to reshape an array with row and col.
arr2 = arr.reshape(3,2)
print(arr2)
```

```python
arr3 = arr.reshape(3,-1)
print(arr3)
 # in large array using -1 to calculate automatically col.
```

```python
print(arr3.ndim)
```

```python
## initializing numpy arrays from a nested python list.
list2 = [1,2,3,4,5]
list3 = [2,3,4,5,6]
list4 = [9,7,6,8,9]

arr4 = np.array([list2,list3,list4])
print(arr4)
```

```python
arr4.tolist()
```

```python
print(arr4.shape)
```

```python
arr4.reshape(1,15)
```

```python
# Numpy-Attributes_ [[]]
a = np.array([[1,2,3],[4,5,6]])
print(a.shape)
```

```python
# reshape the n dim array.
a.shape = (3,2)
print(a)
```

```python
# to create a sequence of values and then created as an array using range fn.
r = range(24)
print(r)
```

```python
# to creat a sequence of values, use arange.
a = np.arange(24)
print(a)
print(a.ndim)
```

```python
# to reshaping the array 'a'.
b = a.reshape(6,4,1)
print(b)
```

```python
# NumPy Arithmetic Operations.
x = np.array([[1,2],[3,4]], dtype=np.float64)
y = np.array([[5,6],[7,8]], dtype=np.float64)
print(x)
print(y)
print(x+y) # print(np.add(x+y))
```

```python
print(x-y)
print(np.subtract(x,y))
```

```python
print(x*y)
print(np.multiply(x,y))
```

```python
print(x/y)
print(np.divide(x,y))
```

```python
print(x.dot(y)) # rows(x)*cols(y)
```

```python
# numpy.sum()- returns sum of all array elements over a given axis.
print(np.sum(x))
```

```python
# sum of each column.(axis=0; performing sum op. on col. and axis=1; sum of rows.)
print(np.sum(x, axis=0))
```
