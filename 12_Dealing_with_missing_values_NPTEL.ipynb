```python
import pandas as pd
```

```python
import numpy as np
```

```python
df = pd.read_csv('/content/drive/MyDrive/Toyota.csv', index_col=0, na_values=['??','????'])
df.shape
```

```python
# Making copies of df for further operations.

df1 = df.copy()
```

```python
df2 = df1.copy()
```

```python
df3 = df1.copy()
```

```python
# To check null values in pd- isnull() and isna() are used, it gives boolean value.
# How many no. of rows having missing columns.

df.isna().sum()
```

```python
# To know where the missing value is located.

missing = df2[df2.isnull().any(axis=1)]
missing
```

```python
# describe() fn. Tto know whether numerical variables should be imputed with mean or median.

df2.describe()
```

```python
# since there is no much difference b/w the mean and median, we are going to impure the age variable with the mean value.

df2['Age'].mean()
```

```python
# now we're going to use 55.67 in the missing value, we use fillna() operation.

df2['Age'].fillna(df2['Age'].mean(), inplace=True)
df2
```

```python
# Missing values of 'KM'.

df2['KM'].median()
```

```python
df2['KM'].fillna(df2['KM'].median(), inplace=True)
df2
```

```python
# Missing values of 'HP'.

df2['HP'].mean()
```

```python
df2['HP'].fillna(df2['HP'].mean(), inplace=True)
df2
```

```python
# To check missing value after filling values.

df2.isnull().sum()
```

```python
# series.value_counts() returns a series containing counts of missing values,
# The values will be in descending order so that 1st element is the most frequently-occurring element.
# Excludes NA values by default.

df2['FuelType'].value_counts()
```

```python
# To get the model value of FuelType for filling values.

df2['FuelType'].value_counts().index[0]
```

```python
df2['FuelType'].fillna(df2['FuelType'].value_counts().index[0])
```

```python
# To get the mode value of Metcolor.

df2['MetColor'].mode()
```

```python
# To fill the mode in Metcolor with indexing.

df2['MetColor'].fillna(df2['MetColor'].mode()[0])
df2
```

```python
# To check the missing value in df.

df2.isnull().sum()
```

```python
# Imputing Missing Values Using lambda functions.
# To fill the NA/NaN values in both numerical and categorical variables at one stretch.
# applying on third copy of df.

def impute_values(x):
  if x.dtype == 'float' or x.dtype == 'int':
    return x.fillna(x.mean())
  else:
    return x.fillna(x.mode()[0])
df3 = df3.apply(impute_values)
df3
```

```python
# To check the missing values.

df3.isnull().sum()
```
