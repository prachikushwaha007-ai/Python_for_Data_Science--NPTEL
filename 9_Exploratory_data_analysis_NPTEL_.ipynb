```python
df = pd.read_csv('/content/drive/MyDrive/Toyota.csv', index_col=0, na_values=["??","???"])
df
```

```python
df1 = df.copy()  # to make an copy of df for analysis.
df1
```

```python
# pandas.crosstab() to compute a frequency table of the factors.
pd.crosstab(index=df1['FuelType'], columns='count', dropna=True)
```

```python
# two way table - pandas.crosstab()
pd.crosstab(index = df1['Automatic'], columns = df1['FuelType'], dropna=True)
```

```python
# Two-way table-joint probability, we use normalize = True for no. to proportion.
pd.crosstab(index = df1['Automatic'], columns = df1['FuelType'], normalize = True, dropna=True)
```

```python
# two - way table- marginal probability use margins = True.
pd.crosstab(index = df1['Automatic'], columns = df1['FuelType'], margins = True, normalize = True, dropna=True)
```

```python
# conditional probability.
pd.crosstab(index = df1['Automatic'], columns = df1['FuelType'], normalize = 'index', dropna=True)
```

```python
pd.crosstab(index = df1['Automatic'], columns = df1['FuelType'], normalize = 'columns', dropna=True)
```

```python
# coorelation to compute pairwise correlatin of col. excluding NA/values
# df.corr(self, method='pearson')
numerical_data = df1.select_dtypes(include=['number'])
print(numerical_data.shape)
```

```python
# correlation matrix
corr_matrix = numerical_data.corr()
corr_matrix
```
