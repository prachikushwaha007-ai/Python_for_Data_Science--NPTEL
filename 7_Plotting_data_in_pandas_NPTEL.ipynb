```python
import pandas as pd
```

```python
url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
```

```python
df = pd.read_csv(url)
```

```python
df.head(891)
```

```python
df["Sex"].value_counts()
```

```python
# plotting bar graph
df["Sex"].value_counts().plot.bar()
```

```python
df["Sex"].value_counts().plot.bar(title="Passengers on the Titanic")
```

```python
df["Sex"].value_counts().plot.bar(title="Passengers on the Titanic", color=["Blue","Red"])
```

```python
# for Horizontal graph
df["Sex"].value_counts().plot.barh(title="Passengers on the Titanic", color=["blue","red"])
```

```python
df["Age"].value_counts().plot.bar(title="Age of passenges on Titanic")
```

```python
#to make a pie chart
df["Sex"].value_counts().plot.pie(title= "passenges on Titanic")
```

```python
# To change size of graph
df["Sex"]. value_counts().plot.pie(figsize=(10,6))
```

```python
df["Sex"].value_counts(). plot.pie(labels=["Male","Female"])
```

```python
#To draw percentage in a graph.
df["Sex"].value_counts(). plot.pie(labels=["Male","Female"],autopct="%.2f")
```

```python
#to make a pie chart
df["Age"].value_counts().plot.pie(title= "passenges on Titanic")
```

```python
# To plot scatter graph.
df.plot.scatter(x="Age", y="Fare")
```

```python
# specify fare price with age and classes.
df.plot.scatter(x="Age",y="Fare",c="Pclass",cmap="viridis")
```

```python
# for enhance df in a "classes".
df.loc[(df.Pclass == 1), "Pclass"] = "First"
df.loc[(df.Pclass == 2), "Pclass"] = "Second"
df.loc[(df.Pclass == 3), "Pclass"] = "Third"
df
```

```python
# To change color.
df.loc[(df.Pclass == "first"), "Pclass"] = "Red"
df.loc[(df.Pclass == "second"), "Pclass"] = "Green"
df.loc[(df.Pclass == "third"), "Pclass"] = "purple"
df.plot.scatter(x="Age",y="Fare",c="Pclass")
```

```python
#for plotting age graph
df["Pclass"] = df.Pclass.astype("category")
df
```

```python
df.plot.scatter(x="Age", y="Fare", c="Pclass",cmap="viridis")
```

```python
#For enhancing the graph size.
df.plot.scatter(x="Age", y="Fare", c="Pclass",cmap="viridis", figsize= (15,5))
```
